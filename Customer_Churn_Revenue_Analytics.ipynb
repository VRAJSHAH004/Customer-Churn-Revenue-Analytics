```python
# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Check missing values created after conversion
print("Missing Values in TotalCharges:", df["TotalCharges"].isnull().sum())

# Remove rows with missing TotalCharges
df = df.dropna(subset=["TotalCharges"])

# Verify the data type
print("\nUpdated Data Type:")
print(df["TotalCharges"].dtype)

# Verify dataset information
print("\nUpdated Dataset Info:")
df.info()
```

##**Phase - 3 Business KPI Analysis**

###Calculate KPI

```python
# Total Customers
total_customers = len(df)

# Total Revenue
total_revenue = df["TotalCharges"].sum()

# Churned Customers
churned_customers = (df["Churn"] == "Yes").sum()

# Active Customers
active_customers = (df["Churn"] == "No").sum()

# Churn Rate
churn_rate = (churned_customers / total_customers) * 100

# Retention Rate
retention_rate = (active_customers / total_customers) * 100

# Average Monthly Charges
avg_monthly = df["MonthlyCharges"].mean()

# Average Total Charges
avg_total = df["TotalCharges"].mean()

# Average Tenure
avg_tenure = df["tenure"].mean()

# Average Revenue per Customer
avg_revenue = total_revenue / total_customers
```

```python
print("========== BUSINESS KPIs ==========")

print(f"Total Customers          : {total_customers}")
print(f"Total Revenue            : ${total_revenue:,.2f}")
print(f"Churned Customers        : {churned_customers}")
print(f"Active Customers         : {active_customers}")
print(f"Churn Rate               : {churn_rate:.2f}%")
print(f"Retention Rate           : {retention_rate:.2f}%")
print(f"Average Monthly Charges  : ${avg_monthly:.2f}")
print(f"Average Total Charges    : ${avg_total:.2f}")
print(f"Average Tenure           : {avg_tenure:.2f} Months")
print(f"Average Revenue/Customer : ${avg_revenue:.2f}")
```

##**Phase - 4 Customer Behavior Analysis**

##Chart 1 - Churn by Contract

```python
import matplotlib.pyplot as plt

contract = pd.crosstab(df["Contract"], df["Churn"])

ax = contract.plot(
    kind="bar",
    stacked=True,
    figsize=(8,5)
)

plt.title("Customer Churn by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.legend(title="Churn")
for container in ax.containers:
    labels = [
        f"{int(v)}" if v >= 100 else ""
        for v in container.datavalues
    ]
    ax.bar_label(container, labels=labels, fontsize=9)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

###Business Insight:

Customers with Month-to-month contracts experience the highest churn, while customers with One-year and Two-year contracts show much higher retention. Longer-term contracts improve customer loyalty.

##Chart 2 — Churn by Payment Method

```python
payment = pd.crosstab(df["PaymentMethod"], df["Churn"])

ax = payment.plot(
    kind="barh",
    figsize=(10,5)
)

plt.title("Customer Churn by Payment Method")
plt.xlabel("Number of Customers")
plt.ylabel("Payment Method")

for container in ax.containers:
    ax.bar_label(container, fontsize=9)

# Move legend outside the chart
plt.legend(
    title="Churn",
    loc="upper left",
    bbox_to_anchor=(1.02, 1)
)

plt.tight_layout()
plt.show()
```

###Business Insight:
Customers using Electronic Check have the highest churn compared to other payment methods. Customers using automatic payment methods tend to remain with the company longer.

##Chart 3 – Churn by Internet Service

```python
internet = pd.crosstab(df["InternetService"], df["Churn"])

ax = internet.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Customer Churn by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

# Add labels
for container in ax.containers:
    ax.bar_label(container, fontsize=9)

plt.legend(
    title="Churn",
    loc="upper left",
    bbox_to_anchor=(1.02, 1)
)

plt.tight_layout()
plt.show()
```

###Business Insight:
Customers using Fiber Optic internet service have the highest churn, while customers with DSL or no internet service show lower churn levels. Service quality or pricing for Fiber Optic customers may require improvement.

##Chart 4 – Churn by Gender

```python
gender = df.groupby("gender")["Churn"].apply(lambda x: (x=="Yes").sum())

plt.figure(figsize=(6,6))

plt.pie(
    gender,
    labels=gender.index,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.80,
    wedgeprops=dict(width=0.45)
)

plt.title("Customer Churn by Gender")

plt.tight_layout()
plt.show()
```

###Business Insight:
Customer churn is distributed almost equally between male and female customers, indicating that gender has little influence on customer churn.

##Chart 5 – Churn by Senior Citizen

```python
senior = df.groupby("SeniorCitizen")["Churn"].apply(lambda x: (x=="Yes").sum())

labels = ["Non-Senior", "Senior"]

plt.figure(figsize=(6,6))

plt.pie(
    senior,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.75
)

plt.title("Customer Churn by Senior Citizen")

plt.tight_layout()
plt.show()
```

###Business Insight:
Senior citizens account for a higher proportion of churn relative to their customer base, suggesting they may require additional support, personalized services, or targeted retention programs.

##Chart 6 — Customer Churn by Tenure Group

```python
bins = [0, 12, 24, 36, 48, 60, 72]

labels = [
    "0-12",
    "13-24",
    "25-36",
    "37-48",
    "49-60",
    "61-72"
]

df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=bins,
    labels=labels
)

tenure = (
    df[df["Churn"] == "Yes"]
    .groupby("TenureGroup", observed=False)
    .size()
)

plt.figure(figsize=(8,5))

plt.plot(
    tenure.index.astype(str),
    tenure.values,
    marker="o",
    linewidth=2
)

# Add labels above points
for x, y in zip(tenure.index.astype(str), tenure.values):
    plt.text(x, y + 15, str(y), ha="center", fontsize=9)

plt.title("Customer Churn by Tenure Group")
plt.xlabel("Tenure Group (Months)")
plt.ylabel("Number of Churned Customers")

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

###Business Insight:
Customers with shorter tenure (0–12 months) experience the highest churn, while customers with longer tenure are generally more loyal and less likely to leave.

##**Phase - 5 Revenue Analysis**

```python
# Create Tenure Group
bins = [0, 12, 24, 36, 48, 60, 72]
labels = ["0-12", "13-24", "25-36", "37-48", "49-60", "61-72"]

df["TenureGroup"] = pd.cut(df["tenure"], bins=bins, labels=labels)

# Create Customer Segment
df["Segment"] = df["SeniorCitizen"].map({
    0: "Non-Senior",
    1: "Senior"
})


# ==========================================
# Revenue by Contract
# ==========================================

print("=" * 70)
print("Revenue Analysis by Contract")
print("=" * 70)

rev_contract = (
    df.groupby("Contract")["TotalCharges"]
      .sum()
      .sort_values(ascending=False)
      .reset_index(name="Total Revenue")
)

rev_contract["Total Revenue"] = rev_contract["Total Revenue"].map(lambda x: f"${x:,.2f}")

print(rev_contract.to_string(index=False))

print("\nBusiness Insight:")
print("Long-term (Two-Year) contracts generate the highest revenue,")
print("showing that loyal customers contribute the greatest business value.\n")


# ==========================================
# Revenue by Payment Method
# ==========================================

print("=" * 70)
print("Revenue Analysis by Payment Method")
print("=" * 70)

rev_payment = (
    df.groupby("PaymentMethod")["TotalCharges"]
      .sum()
      .sort_values(ascending=False)
      .reset_index(name="Total Revenue")
)

rev_payment["Total Revenue"] = rev_payment["Total Revenue"].map(lambda x: f"${x:,.2f}")

print(rev_payment.to_string(index=False))

print("\nBusiness Insight:")
print("Electronic Check customers generate the highest revenue.")
print("Combined with previous churn analysis, this segment should")
print("be prioritized for customer retention.\n")


# ==========================================
# Revenue by Internet Service
# ==========================================

print("=" * 70)
print("Revenue Analysis by Internet Service")
print("=" * 70)

rev_service = (
    df.groupby("InternetService")["TotalCharges"]
      .sum()
      .sort_values(ascending=False)
      .reset_index(name="Total Revenue")
)

rev_service["Total Revenue"] = rev_service["Total Revenue"].map(lambda x: f"${x:,.2f}")

print(rev_service.to_string(index=False))

print("\nBusiness Insight:")
print("Fiber Optic customers contribute the highest revenue.")
print("Retaining these customers can significantly protect")
print("overall business revenue.\n")


# ==========================================
# Revenue by Customer Segment
# ==========================================

print("=" * 70)
print("Revenue Analysis by Customer Segment")
print("=" * 70)

rev_segment = (
    df.groupby("Segment")["TotalCharges"]
      .sum()
      .sort_values(ascending=False)
      .reset_index(name="Total Revenue")
)

rev_segment["Total Revenue"] = rev_segment["Total Revenue"].map(lambda x: f"${x:,.2f}")

print(rev_segment.to_string(index=False))

print("\nBusiness Insight:")
print("Non-Senior customers contribute the largest share of")
print("overall revenue, making them the primary revenue")
print("generating customer segment.\n")


# ==========================================
# Revenue by Tenure Group
# ==========================================

print("=" * 70)
print("Revenue Analysis by Tenure Group")
print("=" * 70)

rev_tenure = (
    df.groupby("TenureGroup", observed=True)["TotalCharges"]
      .sum()
      .sort_values(ascending=False)
      .reset_index(name="Total Revenue")
)

rev_tenure["Total Revenue"] = rev_tenure["Total Revenue"].map(lambda x: f"${x:,.2f}")

print(rev_tenure.to_string(index=False))

print("\nBusiness Insight:")
print("Customers with longer tenure generate the highest")
print("revenue, highlighting the importance of long-term")
print("customer retention.\n")
```

##Chart 9 - Correlation Heatmap

```python
corr = df[["tenure", "MonthlyCharges", "TotalCharges"]].corr()

fig, ax = plt.subplots(figsize=(6,5))

im = ax.imshow(corr, cmap="Blues")

ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))

ax.set_xticklabels(corr.columns)
ax.set_yticklabels(corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        ax.text(
            j,
            i,
            f"{corr.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=10,
            color="black"
        )

plt.colorbar(im)
plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

###Business Insight:
Customer tenure has the strongest relationship with Total Charges, indicating that long-term customers contribute the highest lifetime revenue.

##**Phase - 7 Recommendations**

## Business Recommendations

Based on the customer churn and revenue analysis, the following recommendations are suggested to improve customer retention and maximize business revenue.

### 1. Promote Long-Term Contracts
- Encourage Month-to-Month customers to switch to One-Year or Two-Year contracts through discounts, loyalty rewards, or bundled service offers.
- Long-term contracts are associated with lower churn and higher lifetime revenue.

---

### 2. Improve Customer Retention During the First Year
- Focus on customers with 0–12 months of tenure, as they experience the highest churn.
- Strengthen onboarding, customer support, and engagement during the early months.

---

### 3. Enhance Fiber Optic Customer Experience
- Improve service quality, pricing, and customer support for Fiber Optic customers.
- This customer segment generates high revenue but also shows higher churn.

---

### 4. Strengthen Payment Method Retention
- Encourage Electronic Check customers to switch to automatic payment methods through incentives or rewards.
- This may improve customer retention while protecting a valuable revenue segment.

---

### 5. Reward High-Value Customers
- Develop loyalty programs for long-term customers with high Total Charges.
- Personalized offers and exclusive benefits can increase customer lifetime value.

---

### 6. Monitor Customer Churn Continuously
- Track important KPIs such as Churn Rate, Customer Retention Rate, and Revenue on a regular basis.
- Use data-driven insights to identify high-risk customers and take proactive retention actions.

---

## Expected Business Impact

Implementing these recommendations can help the company:

- Reduce customer churn.
- Increase customer retention.
- Improve customer satisfaction.
- Protect high-value customer segments.
- Increase long-term business revenue.
- Support better data-driven business decisions.
